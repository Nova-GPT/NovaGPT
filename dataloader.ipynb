```python
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer
from utils import TrainingData
```

```python
class TextDataset(Dataset):
    def __init__(self, text, tokenizer, block_size=128, train_test_split = 0.9):
        """
        text: full text string
        tokenizer: any tokenizer with .encode()
        block_size: sequence length of training examples
        """
        self.tokenizer = tokenizer
        self.block_size = block_size
        data = torch.tensor(tokenizer.encode(text), dtype=torch.long)
        n = int(train_test_split * len(data))
        self.train_data = data[:n]
        self.val_data = data[n:]
    
    def __len__(self):
        # each example is a block_size chunk
        return len(self.train_data) - self.block_size
    
    def __getitem__(self, idx):
        x = self.train_data[idx : idx + self.block_size]
        y = self.train_data[idx + 1 : idx + 1 + self.block_size]
        return x, y
```
